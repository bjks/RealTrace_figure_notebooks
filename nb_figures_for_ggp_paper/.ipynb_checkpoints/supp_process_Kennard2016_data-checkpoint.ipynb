## First processing and removal of obvious errors in the input file

```python
def read_all_txts(files):
    cells = {}
    for file in files:
        file_prefix = file.split("/")[-1].split("_")[0]
        with open(file, "r", encoding="utf-8") as f:
            for line in f:
                if line.startswith("CELL"):
                    cell_id = file_prefix+"_"+line.strip().replace(":\t", "_")
                    cells[cell_id] = {}
                    current_cell = cell_id
                else:
                    line = line.strip()
                    variable = line.split(":")[0]
                    data = np.array(line.split(":")[1].strip().split("\t")).astype(float)
                    cells[current_cell][variable] = data
    return cells

def filter_extreme_outliers(cells):
    new_cell_dict = {}
    for cell in list(cells.keys()):
        if cells[cell]["L"][0]>cells[cell]["L"][-1]:
            continue
        if np.any(np.abs(np.diff( np.log(cells[cell]["L"]) ))>0.3):
            continue
                    
        if len(cells[cell]["T"])<3:
            continue
        new_cell_dict[cell] = cells[cell]
    return new_cell_dict
    

def filter_nans(cells):
    new_cell_dict = {}
    for cell in list(cells.keys()):
        if not np.all(np.isfinite(cells[cell]["L"])):
            continue
        if not np.all(np.isfinite(np.log(cells[cell]["L"]))):
            continue
        if not np.all(np.isfinite(cells[cell]["F1"])):
            continue
        if not np.all(np.isfinite(np.log(cells[cell]["F1"]))):
            continue
        if np.any(np.diff(cells[cell]["T"])==0):
            continue
        new_cell_dict[cell] = cells[cell]
    return new_cell_dict


def filter_by_r2(cells, r2, only_positive=True):
    new_cell_dict = {}
    for cell in list(cells.keys()):
        r = np.corrcoef(cells[cell]["T"], np.log(cells[cell]["L"]))[0,1]
        if r<0:
            continue
        if r**2 < r2:
            continue
        new_cell_dict[cell] = cells[cell]
    return new_cell_dict
    
def get_n_cells(cells, n=1000):
    new_cell_dict = {}
    for cell in list(cells.keys())[:n]:
        new_cell_dict[cell] = cells[cell]
    return new_cell_dict
    

def r2_hist(cells, title=None):
    fig, ax = plt.subplots(1, 1, figsize=(5, 3))
    ax.set_title(title)
    rs = [np.corrcoef(cells[cell]["T"], np.log(cells[cell]["L"]))[0,1] for cell in list(cells.keys())]
    ax.hist(rs, bins="auto")
    ax.set_yscale("log")
    ax.set_xlabel("Corr(t, log(l)")
    ax.set_xlim(0.5, 1)
    plt.show()
    
def process_raw_data(raw_data, outfile):

    dataset = pd.DataFrame()
    for cell in raw_data.keys():        
        cell_df = pd.DataFrame(raw_data[cell])
        cell_df["cell_id"] = cell
        dataset = pd.concat([dataset, cell_df], ignore_index=True)
        
    dataset["parent_id"] = -1
    dataset.to_csv(outfile)
    return dataset


    
directories = ["../../ecoli_other_labs/Kennard2016/Data_RDM_P5ori/Raw",
               "../../ecoli_other_labs/Kennard2016/Data_CAA_P5ori/Raw",
               "../../ecoli_other_labs/Kennard2016/Data_Glc_P5ori/Raw"]

data_sets =[]
raw_datas =[]
for directory in directories:
    filenames = get_suffix_files(directory, suffix='txt') 
    raw_data = read_all_txts( filenames)
    print(directory.split("/")[-2], "before filter", len(raw_data))
    
    raw_data = filter_nans(raw_data)
    print("after NAN filter", len(raw_data))

    
    raw_data = filter_extreme_outliers(raw_data)
    print("after outlier filter", len(raw_data))
    
    # r2_hist(raw_data, title =directory.split("/")[-2])
    # raw_data = filter_by_r2(raw_data, 0.9, only_positive=True)
    # print("after R2 filter", len(raw_data))

    raw_data = get_n_cells(raw_data, n=1000)
    raw_datas.append(raw_data)
    outfile = directory.split("/Raw")[0] + "_processed.csv"
    dataset = process_raw_data(raw_data, outfile)
    data_sets.append(dataset)

```

```python
def plot_growth_hist(df):
    fig, ax = plt.subplots(1, 1, figsize=(5, 3))
    # ax.set_title(title)
    ax.hist(df["mean_l"], bins="auto")
    mean = np.mean(df["mean_l"])
    sd = np.std(df["mean_l"])
    ax.axvline(mean-4*sd)
    ax.axvline(mean+4*sd)
    
    ax.set_yscale("log")
    ax.set_xlabel("Growth")
    # ax.set_xlim(0.5, 1)
    plt.show()

def filter_cells_by_growth_outlier(df, n_std):
    mean = np.mean(df["mean_l"])
    sd = np.std(df["mean_l"])
    lb = mean - n_std*sd
    ub = mean + n_std*sd    
    too_low =  df[df["mean_l"]<lb]["cell_id"]
    too_high = df[df["mean_l"]>ub]["cell_id"]
    cell_ids = np.unique(np.append(too_low, too_high))
    print(len(cell_ids))
    
    new_df = df[~df["cell_id"].isin(cell_ids)] 
    print(len(np.unique(new_df["cell_id"])))
    
    new_df["L"] = np.exp(new_df["log_length"])
    new_df["T"] = new_df["time"]
    # new_df["F1"] = new_df["fp"]
    new_df["dummy_gfp"] = np.random.normal(loc=0, scale=1e-3, size=len(new_df["time"]))
    
    return new_df[["cell_id", "parent_id", "L", "T", "dummy_gfp"]]

    
for pf in prediction_files:
    print(pf)
    prediction_processed_df = pd.read_csv(pf, skiprows=13)
    plot_growth_hist(prediction_processed_df)
    new_out = os.path.join(*pf.split("/")[:-2]) + "/" + pf.split("/")[-1].split("_pro")[0] +"_cleaned.csv"
    print(new_out)
    filter_cells_by_growth_outlier(prediction_processed_df, 4.).to_csv(new_out)
```

```python
### Get rate stats ###

def calc_cv(means, covs=None):
    if np.any(covs==None):
        return np.std(means)/np.mean(means)
    else:
        return np.sqrt(np.sum(np.power(means, 2) + covs)/len(means) - np.mean(means)**2)/np.mean(means)

def calc_std(means, covs=None):
    if np.any(covs==None):
        return np.std(means)
    else:
        return np.sqrt(np.sum(np.power(means, 2) + covs)/len(means) - np.mean(means)**2)
    
def calc_mean(means, covs=None):
    if np.any(covs==None):
        mean = np.mean(means)
    else:
        mean = np.sum(means/covs) /np.sum(1/covs)
    return mean

def estimate_rate_stats(filename):
    
    data_prediction = pd.read_csv(filename, skiprows=header_lines(filename))
    cells_prediction = ggp_df2cells(data_prediction)
    cell_stats = {}
    for i in ["l", "q"]:
        zi = "{:s}t".format(i)
        zi_cov = "cov_{:s}{:s}".format(i, i)

        zi_coll_prediction = np.concatenate([cell.__dict__[zi] for cell in cells_prediction])
        zi_coll_prediction_err = np.concatenate([cell.__dict__[zi_cov] for cell in cells_prediction])
        cell_stats["{:s}_mean".format(i)] = calc_mean(zi_coll_prediction, zi_coll_prediction_err)
        cell_stats["{:s}_cv".format(i)] = calc_cv(zi_coll_prediction, None)
        cell_stats["{:s}_std".format(i)] = calc_std(zi_coll_prediction, zi_coll_prediction_err)

    cell_stats["q_cv/l_cv"]= cell_stats["q_cv"]/ cell_stats["l_cv"]
    ## Adj. units lambda in 1/h, q in 1000/h
    cell_stats['l_mean']*=60
    cell_stats['l_std']*=60
    
    cell_stats['q_mean']*=60/1000
    cell_stats['q_std']*=60/1000

    sample = "_".join(filename.split("/")[-1].split(".")[0].split("_")[:3])
    cell_stats["sample"]= sample
    
    cell_stats["condition"] = filename.split("/")[-1].split("_")[0]
    cell_stats["promoter"] = filename.split("/")[-1].split("_")[1]
    return cell_stats


all_files_dany = sorted(find_files("../../data_realtrace_paper/ecoli_data/", ending="prediction.csv"))

promoters_to_plot = ['hi1', 'hi3', 'rplN', 'rrnB']
conditions_to_plot = conditions[:3]

all_files_dany = [file for file in all_files_dany if file.split("/")[-1].split("_")[0] in conditions_to_plot and file.split("/")[-1].split("_")[1] in promoters_to_plot]        
all_files_dany = sort_by_condition(all_files_dany)



cell_stats = {}
for file in prediction_files_cleaned :
    data_set_name = file.split("/")[-1].split("_rawdata")[0]
    print("|", end="")
    cell_stats[data_set_name] = estimate_rate_stats(file)
    cell_stats[data_set_name]["source"] = "Kennard et al. 2016"


for file in  all_files_dany :
    data_set_name = file.split("/")[-1].split("_rawdata")[0]
    print("|", end="")
    cell_stats[data_set_name] = estimate_rate_stats(file)
    cell_stats[data_set_name]["source"] = "this study"
```
