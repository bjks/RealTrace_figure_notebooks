```python
def df2cells_2ou(dataset, time="time_min", 
            log_length="log_length", gfp="gfp", 
            cell_id="cell_id", parent_id="parent_id", 
            lt="lt", qt="qt", lane=None):
    """ 
    dataset (pandas data frame as read from csv file) to list of Cell instances 
    either using columns with measurment noise or without
    """
    cell_list = []
    last_cell = ""
    for _, row in dataset.iterrows(): 
        if row[cell_id] != last_cell:
            
            if lane !=None:
                c = str(row[lane])+'.'+str(row[cell_id])
                p = str(row[lane])+'.'+str(row[parent_id])
            else:
                c = str(row[cell_id])
                p = str(row[parent_id])
                
            p = row[parent_id]

            lambda0 = row[lt]
            q0 = row[qt]

            new_cell = Cell(row[log_length], row[gfp], 
                        lambda0, q0, 
                        time0=row[time],
                        cell_id=c, 
                        parent_id=p)
            
            new_cell.lt1 = [row["lt1"]]
            new_cell.lt2 = [row["lt2"]]

            new_cell.qt1 = [row["qt1"]]
            new_cell.qt2 = [row["qt2"]]
            
            
            cell_list.append(new_cell)
        else:
            cell_list[-1].log_length.append(row[log_length])
            cell_list[-1].gfp.append(row[gfp])
            cell_list[-1].time.append(row[time])

            if lt != None: 
                cell_list[-1].lt.append(row[lt])
                cell_list[-1].lt1.append(row["lt1"])
                cell_list[-1].lt2.append(row["lt2"])
                cell_list[-1].qt1.append(row["qt1"])
                cell_list[-1].qt2.append(row["qt2"])
                
            if qt != None:
                cell_list[-1].qt.append(row[qt])

        last_cell = row[cell_id]
    return cell_list

full_data = pd.read_csv(directory + sample + "_full.csv", 
                       dtype={"parent_id":str, "cell_id":str})
cells_full = df2cells_2ou(full_data, time="time_min")
```

```python

def get_path_idx(cells, idx0=0):
    idx = []
    last_cell = cells[idx0].parent_id
    for i, _ in enumerate(cells):
        if last_cell == cells[i].parent_id: 
            last_cell = cells[i].cell_id
            idx.append(i)
    return idx

def gaussian(x, y):
    mean = np.mean(y)
    sigma = np.std(y)
    return 1./(np.sqrt(2*np.pi)*sigma) * np.exp(-(x-mean)**2/(2*sigma**2))


def gauss_function(x, a, x0, sigma):
    return a*np.exp(-(x-x0)**2/(2*sigma**2))
```
