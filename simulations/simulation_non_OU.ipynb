```python

# ========== Simulation parameters ========== #
save_dataset = True    # saves data set as csv
run_ggp = True         # runs the ggp code (located in "../bin/" relative to this nb)

dt = 0.01
dt_measument = 3 # in minutes

n_cells = 1000 # number of cells that will be simulated
discard_cells = 20
lanes = 1

log_length0 = np.log(2)
gfp0 = 24975.8


div_mode = "adder"
division_log_length = None   # for sizer: division, when log_length hits division_log_length
division_time = None          # for timer: division, when cell cycle time hits division_time
division_addition = log_length0       # for adder: divsion, when division_addition in log_length was added in cell cycle


data_set_name = "non_OU"
# taken from glucose_hi1_f01234578910_b_prediction in data_dany_old_data_output_20221215
parameter_set  =  {"mean_lambda": 0.0114137, 
                    "gamma_lambda": 0.166994,
                    "gamma_lambda1": 0.166994,
                    "gamma_lambda2": 0.166994  /10.,
                    "var_lambda": 4.9481e-06  ,
                    "var_lambda1": 4.9481e-06/4.  ,
                    "var_lambda2": 4.9481e-06  /40.,
                   #
                    "mean_q": 118.524,
                    "gamma_q": 0.0595599,
                    "gamma_q1": 0.0595599,
                    "gamma_q2": 0.0595599 /10.,
                    "var_q": 430.055,
                    "var_q1": 430.055 /4.,
                    "var_q2": 430.055     /40.,
                   # 
                    "beta": 0,
                    "var_x": 0.000136701,
                    "var_g": 12.255,
                    "var_dx": 0.00468653,
                    "var_dg": 1.53195}
```

```python
def growth_2ou(cell, dt, lt1, lt2):  

    lt = lt1+lt2

    # calculate next step
    next_step = cell.log_length[-1] + lt*dt

    # save everything
    cell.lt.append(lt)
    cell.lt1.append(lt1)
    cell.lt2.append(lt2)
    cell.log_length.append(next_step)
    cell.length.append(np.exp(next_step))
    return cell


def gfp_production_2ou(cell, dt, qt1, qt2, beta):
    # calculate next step
    qt = qt1+qt2
    next_step = cell.gfp[-1] + cell.length[-1]*qt*dt - cell.gfp[-1]*beta*dt   

    cell.qt.append(qt)
    cell.qt1.append(qt1)
    cell.qt2.append(qt2)
    cell.gfp.append(next_step)
    return cell 
    
def cell_division(cell, var_dx, var_dg, no_cells, cell_division_model, dt):
    if cell_division_model=="binomial":
        x = cell.log_length[-1]
        g = cell.gfp[-1]
        x_d1 = np.random.normal(loc=x-np.log(2), scale=np.sqrt(var_dx)) 
        if var_dg==0:
            g_d1 = g * np.exp(x_d1 - x)
        else:
            if g <= 0:
                g_d1 = 0
            else:   
                g_d1 = np.random.binomial(g/var_dg, np.exp(x_d1 - x))*var_dg
                if g_d1 < 1:
                    g_d1 =  10

        x_d2 = np.log(np.exp(x)-np.exp(x_d1))
        
        if g <= 0:
            g_d2 = 0
        else:
            g_d2 = g - g_d1

        cell1 = Cell(x_d1,g_d1, cell.lt[-1],  
                     cell.qt[-1],
                        time0 = cell.time[-1]+dt,
                        cell_id = no_cells + 1, parent_id=cell.cell_id)
        cell1.lt1 = [cell.lt1[-1]]
        cell1.lt2 = [cell.lt2[-1]]
        cell1.qt1 = [cell.qt1[-1]]
        cell1.qt2 = [cell.qt2[-1]]

        cell2 = Cell(x_d2,g_d2, cell.lt[-1], 
                     cell.qt[-1], 
                        time0 = cell.time[-1]+dt,
                        cell_id = no_cells + 2, parent_id=cell.cell_id)
        cell2.lt1 = [cell.lt1[-1]]
        cell2.lt2 = [cell.lt2[-1]]
        cell2.qt1 = [cell.qt1[-1]]
        cell2.qt2 = [cell.qt2[-1]]

    no_cells += 2 
    return cell1, cell2, no_cells

def simulate_cells_ou2(dt, n_cells, parameter_set, div_mode, 
                        division_log_length=None, 
                        division_time=None, 
                        division_addition=None, 
                        log_length0=None,
                        gfp0=None, 
                        tree=True,
                        tmax=np.inf, 
                        verbose=False, 
                        discard_cells=0, 
                        cell_division_model="binomial"):
    if gfp0 == None:
        gfp0 = 3*parameter_set['mean_q']/parameter_set['mean_lambda']
    if log_length0 == None:
        log_length0 = division_log_length-np.log(2)

    cell_queue = [Cell(log_length0, gfp0, parameter_set['mean_lambda'], parameter_set['mean_q'])]
    for i, cell in enumerate(cell_queue):
        cell_queue[i].lt1 = [0]
        cell_queue[i].lt2 = [0]
        cell_queue[i].qt1 = [0]
        cell_queue[i].qt2 = [0]
    
    cells_simulated = []
    no_cells = 0   # total number of cells (in queue and calculated)
    t_total = 0    # the largest t in the current set of cells

    while len(cells_simulated) < n_cells and t_total < tmax:
        if len(cells_simulated)>0:
            idx = get_current_leafs_idx(cell_queue, cells_simulated)
            cell_index = np.random.choice(idx)
        else:
            cell_index = 0

        # print_simulated(cells_simulated)
        # print_queue(cell_queue,cell_index)

        # simulate a random cell in queue
        cell = copy.deepcopy(cell_queue[cell_index])

        # --------------------------------------------------------------- #
        # Simulation on single cell level
        # --------------------------------------------------------------- #
        while True:
            cell.time.append(cell.time[-1]+dt)

            t_total = np.max([t_total, cell.time[-1]])
            if t_total > tmax:
                break

            qt1 = single_ou_step(dt,   parameter_set['mean_q'], 
                                        parameter_set['gamma_q1'], 
                                        parameter_set['var_q1'], 
                                        cell.qt1[-1]) 

            qt2 = single_ou_step(dt,   0, 
                                        parameter_set['gamma_q2'], 
                                        parameter_set['var_q2'], 
                                        cell.qt2[-1]) 
            
            cell = gfp_production_2ou(cell, dt, qt1, qt2, parameter_set['beta'])

            lt1 = single_ou_step(dt,  parameter_set['mean_lambda'], 
                                            parameter_set['gamma_lambda1'], 
                                            parameter_set['var_lambda1'], 
                                            cell.lt1[-1]) 
            lt2 = single_ou_step(dt,  0, 
                                 parameter_set['gamma_lambda2'], 
                                 parameter_set['var_lambda2'], 
                                cell.lt2[-1]) 
            
            cell = growth_2ou(cell, dt, lt1, lt2)
            

            if is_cell_division(cell, div_mode, division_log_length, division_time, division_addition):
                # save the simulated cell
                if discard_cells==0:
                    cells_simulated.append(cell)
                else:
                    discard_cells -= 1
                
                # calc. new init conditions for 2 daughter cells
                cell1, cell2, no_cells = cell_division(cell, parameter_set['var_dx'], 
                                                            parameter_set['var_dg'], 
                                                            no_cells, cell_division_model, dt)

                # remove the simulated cell from queue and add the new ones 
                cell_queue.pop(cell_index)
                cell_queue.append(cell1)
                if tree and discard_cells==0:
                    cell_queue.append(cell2)
                break
            else:
                pass

        if verbose:
            progress_bar_n = np.around(len(cells_simulated)/n_cells*20).astype(int)
            progress_bar = '='*progress_bar_n + ' '*(20-progress_bar_n)
            print("\r|", progress_bar,  "| Progress {:3.0f}%".format(len(cells_simulated)/n_cells*100), " No of cells: ", len(cells_simulated), end='')  
            print('')

    return cells_simulated


# --------- SIMULATE --------- #
cells_simulated = simulate_cells_ou2(dt, n_cells, parameter_set, div_mode,
                                division_log_length, 
                                division_time, division_addition, 
                                log_length0=log_length0, gfp0=gfp0, 
                                tree=True,
                                discard_cells=discard_cells)
```

```python
def cell_to_df_2OU(self, n=1, start=0):
        df = pd.DataFrame({   "cell_id": ([self.cell_id]*len(self.time))[start::n],
                                "time_min": self.time[start::n],
                                "parent_id": ([self.parent_id]*len(self.time))[start::n],
                                "log_length": self.log_length[start::n], 
                                "gfp": self.gfp[start::n],
                                "lt": self.lt[start::n],
                                "lt1": self.lt1[start::n],
                                "lt2": self.lt2[start::n],
                                "qt": self.qt[start::n],
                                "qt1": self.qt1[start::n],
                                "qt2": self.qt2[start::n],
                          }
                         )
        if len(self.segment)>0:
            df['segment']=self.segment[start::n]
        return df

def build_data_set_scale_gfp_noise(cells_simulated, var_x, var_g, dt, atol=1e-4):

    dataset = pd.DataFrame()
    for i in range(len(cells_simulated)):
        next_celldf = cell_to_df_2OU(cells_simulated[i])
        
        next_celldf['log_length_noise'] = next_celldf['log_length'] + np.random.normal(loc=np.zeros_like( next_celldf['log_length']), scale=np.sqrt(var_x))
        next_celldf['gfp_noise'] = next_celldf['gfp'] + np.random.normal(loc=np.zeros_like( next_celldf['gfp']), 
                                                                        scale=np.sqrt(var_g * np.abs(next_celldf['gfp'])))
        dataset = pd.concat([dataset, next_celldf], ignore_index=True)
    
    return dataset[ np.abs( (dataset['time_min'] + atol/2) % dt) < atol ] 



base_dir ='../../data_realtrace_paper/{:s}_simulation{:d}'
out_dir = mk_missing_dir(base_dir.format(data_set_name, n_cells))

# ----------- SAVE ----------- #
dt_measument = 3 # in minutes

dataset = build_data_set_scale_gfp_noise(cells_simulated, 
                                                parameter_set['var_x'], 
                                                parameter_set['var_g'], 
                                                dt_measument,
                                                atol=dt/10)


dataset["length_noise"] = np.exp(dataset["log_length_noise"])
directory, filename = get_next_file_name(out_dir)

write_csv_config(os.path.join(directory, "csv_config.txt"))

write_param_file(os.path.join(directory, "parameters_sim.txt"), parameter_set) 
param_file = write_param_file_noise(os.path.join(directory, "parameters_noise.txt"), 
                                                parameter_set, 
                                                rel_step=0.1, 
                                                log_noise=0.1, fixed=["beta"]) 
dataset.to_csv(filename)




dataset_full = build_data_set_scale_gfp_noise(cells_simulated[:50], 
                                                parameter_set['var_x'], 
                                                parameter_set['var_g'], 
                                                dt,
                                                atol=dt/10)
filename_full = filename[:-4] + "_full.csv"
dataset_full["length_noise"] = np.exp(dataset_full["log_length_noise"])
dataset_full.to_csv(filename_full)
```
